- ### **Read data, Hint: read about “sep” parameter in `read_csv()`.**


```python
import pandas as pd
titanic = pd.read_csv('train.csv')
titanic.head()  
```

- ### **Try to use info, describe, group by to help you analyze the data.**

```python
titanic.info()
titanic.describe()
print("\n the percent survive for sex :")
print(titanic.groupby('Sex')['Survived'].mean())
print("\nthe percent survive for pclass")
print(titanic.groupby('Pclass')['Survived'].mean())
```

- ### **Deal with duplicates if exist.**

```python
print(f"عدد المكررات: {titanic.duplicated().sum()}")
titanic = titanic.drop_duplicates()
print(f"عدد الصفوف بعد حذف المكررات: {len(titanic)}")
```

- ### **Handling missing values and why you used this approach?**

```python
print("القيم المفقودة قبل المعالجة:")
print(titanic.isnull().sum())
titanic['Age'].fillna(titanic['Age'].mean(), inplace=True)
titanic['Cabin'].fillna('Unknown', inplace=True)
titanic['Embarked'].fillna(titanic['Embarked'].mode()[0], inplace=True)
print("\nالقيم المفقودة بعد المعالجة:")
print(titanic.isnull().sum())
```

- ### **Use drop if needed.**

```python
titanic.drop(['PassengerId', 'Name', 'Ticket'], axis=1, inplace=True)
titanic.head()
```

- ### **Save your modified dataframe.**

```python
titanic.to_csv('titanic_cleaned.csv', index=False)
print(" حفظ البيانات نجح!")
```
